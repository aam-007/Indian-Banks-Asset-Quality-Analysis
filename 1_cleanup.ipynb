```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("masterdataset.csv")

# Remove duplicate side-by-side columns
df = df.loc[:, ~df.columns.str.endswith(".1")]

# Remove repeated header rows inside data
df = df[df["Gross NPA (%)"] != "Gross NPA (%)"]

# Clean percentage columns
percent_cols = [
    "Gross NPA (%)",
    "Net NPA (%)",
    "Provision Coverage Ratio (%)",
    "CAR(%)",
    "Tier 1 Capital"
]

for col in percent_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("%", "", regex=False)
        .str.strip()
    )

# Clean Net Profit column
df["Net Profit (cr)"] = (
    df["Net Profit (cr)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

# Convert to numeric
numeric_cols = percent_cols + ["Net Profit (cr)"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Clean Year column
df["Year"] = df["Year"].astype(str).str.strip()

# Drop empty rows and reset index
df = df.dropna(how="all").reset_index(drop=True)

# Save cleaned dataset
df.to_csv("cleaned_masterdataset.csv", index=False)

print("Dataset cleaned successfully.")
print(df.info())
print(df.head())
```
